```python
from Bio import SeqIO
import pandas as pd

def compare_fastas(fasta1_path, fasta2_path, output_csv_path):
    # Lê os IDs das sequências de cada arquivo
    fasta1_records = {record.id for record in SeqIO.parse(fasta1_path, "fasta")}
    fasta2_records = {record.id for record in SeqIO.parse(fasta2_path, "fasta")}

    # União de todos os IDs
    all_ids = fasta1_records.union(fasta2_records)

    data = []

    for seq_id in sorted(all_ids):
        in_fasta1 = seq_id in fasta1_records
        in_fasta2 = seq_id in fasta2_records

        if in_fasta1 and in_fasta2:
            status = "Ambos"
        elif in_fasta1:
            status = "Somente em Zika479ONE"
        else:
            status = "Somente em Zika170ONE"

        data.append({"Sequence_ID": seq_id, "Status": status})

    # Gerar o CSV
    df = pd.DataFrame(data)
    df.to_csv(output_csv_path, index=False)
    print(f"Arquivo CSV gerado: {output_csv_path}")

    # Contar quantos em cada categoria
    counts = df['Status'].value_counts()
    print("\nResumo dos matches:")
    print(counts)
    print("\nTotal de sequências:", counts.sum())

# Exemplo de uso:
compare_fastas("/home/joh/Documentos/Bioinfo - IC/git/BioComp_UFF/data/Zika479ONE/dataset_final.fasta", "/home/joh/Documentos/Bioinfo - IC/git/BioComp_UFF/data/Zika170ONE/dataset_final.fasta", "resultado_comparacao2.csv")
```
